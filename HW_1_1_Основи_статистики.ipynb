**Завдання 5**. Ми провели дослідження віку студентів-першокурсників. Виявилось, що
- 10 студентів мають вік 16 років
- 25 студентів - 17 років
- 20 студентів - 18 років
- 2 студенти - 20 років

Обчислість середнє значення для цієї вибірки будь-яким способом - на чистому Python або з numpy - та виведіть результат на екран.

```python
ages = [16, 17, 18, 20]
counts = [10, 25, 20, 2]

total_age = 0
for i in range(len(ages)):
    total_age = total_age + ages[i] * counts[i]

mean_age = total_age / sum(counts)
print(round(mean_age, 2))
```

```python
mean_age2 = np.average([16, 17, 18, 20], weights=[10, 25, 20, 2])
print(round(mean_age2, 2))
```

**Завдання 6**. Крім даних про суму покупок, ми зібрали ще дані про кількість відвідувань кожного з кожного покупця, які знаходяться в змінній `visits`. Обчисліть коефіцієнт кореляції Пірсона для суми покупок і кількості відвідувань кожним з зазначених методів:  

  6.1. без використання жодних бібліотек, користуючись тільки Python.  
  6.2. використовуючи відповідний метод бібліотеки numpy.  
  
Виведіть результат на екран. Проаналізуйте результат.

Виведіть додатково графік розсіювання викликавши код нижче і зробіть заключення, чи варто щось змінити при підрахунку кореляції, аби коефіцієнт Пірсона показував коректну оцінку лінійної залежності?

```python
visits = [ 6,  2, 10,  3,  9,  9,  2,  3,  6,  6, 10,  2,  5, 10,  3,  5,  4,
        7,  4,  2]
```

```python
n = len(sum_purchases)
mean_x = sum(sum_purchases) / n
mean_y = sum(visits) / n
```

```python
numerator = 0
sum_sq_x = 0
sum_sq_y = 0
for i in range(n):
    dx = sum_purchases[i] - mean_x
    dy = visits[i] - mean_y
    numerator = numerator + dx * dy
    sum_sq_x = sum_sq_x + dx ** 2
    sum_sq_y = sum_sq_y + dy ** 2
```

```python
r_python = numerator / (sum_sq_x * sum_sq_y) ** 0.5
print(round(r_python, 3))
```

```python
r_numpy = np.corrcoef(sum_purchases, visits)[0, 1]
print(round(r_numpy, 3))
```

```python
import matplotlib.pyplot as plt

plt.scatter(sum_purchases, visits)
plt.title('Scatter Plot of Purchases vs. Visits')
plt.xlabel('Sum of Purchases')
plt.ylabel('Number of Visits');
```

```python
clean_purchases = [p for p in sum_purchases if p < 1000]
clean_visits = [v for p, v in zip(sum_purchases, visits) if p < 1000]

r_clean = np.corrcoef(clean_purchases, clean_visits)[0, 1]
print(round(r_clean, 3))
```

Коефіцієнт Пірсона для всіх даних дорівнює −0.30, тобто слабкий від'ємний зв'язок: чим більше відвідувань, тим трохи менша сума покупок. На графіку розсіювання видно дві точки (4310 і 3524 грн), які лежать далеко від решти: 18 клієнтів витратили до 500 грн. Пірсон дуже чутливий до таких викидів, бо вони мають величезні відхилення від середнього і непропорційно впливають на результат.

Якщо прибрати ці дві точки, кореляція стає −0.37: від'ємний зв'язок виявляється сильнішим, ніж здавалося. Отже, для коректної оцінки лінійної залежності викиди варто прибрати перед підрахунком або використати кореляцію Спірмена, яка менш чутлива до викидів. Втім, навіть після очищення зв'язок лишається слабким до помірного, тож робити сильні висновки на 18 спостереженнях не варто.

```python
threshold = np.mean(purchases) + 2 * np.std(purchases, ddof=1)
print(round(threshold, 2))
```

```python
mask = purchases <= threshold

clean_purchases = purchases[mask]
clean_visits = visits_arr[mask]
```

```python
print("Записів до очищення:", len(purchases))
print("Записів після очищення:", len(clean_purchases))
```
